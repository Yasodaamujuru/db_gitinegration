```python
invalid_condition_df = movie_df.withColumn(
    "rejection_reason",

    when(
        col("booking_id").isNull() |
        (trim(col("booking_id")) == ""),
        "Missing Booking ID"
    )

    .when(
        col("customer_id").isNull() |
        (trim(col("customer_id")) == ""),
        "Missing Customer ID"
    )

    .when(
        col("movie_name").isNull() |
        (trim(col("movie_name")) == ""),
        "Missing Movie Name"
    )

    .when(
        col("cinema_name").isNull() |
        (trim(col("cinema_name")) == ""),
        "Missing Cinema Name"
    )

    .when(
        col("city").isNull() |
        (trim(col("city")) == ""),
        "Missing City"
    )

    .when(
        col("screen_type").isNull(),
        "Missing Screen Type"
    )

    .when(
        col("ticket_count").isNull() |
        (col("ticket_count") <= 0),
        "Invalid Ticket Count"
    )

    .when(
        col("ticket_price").isNull() |
        (col("ticket_price") <= 0),
        "Invalid Ticket Price"
    )

    .when(
        col("convenience_fee").isNull() |
        (col("convenience_fee") < 0),
        "Invalid Convenience Fee"
    )

    .when(
        col("booking_date").isNull(),
        "Missing Booking Date"
    )

    .when(
        ~upper(trim(col("booking_status"))).isin(
            "CONFIRMED", "COMPLETED", "CANCELLED"
        ),
        "Invalid Booking Status"
    )
)
```

```python
def check_df_compatibility(target_df, source_df):

    target_schema = {
        field.name: field.dataType.simpleString()
        for field in target_df.schema.fields
    }

    source_schema = {
        field.name: field.dataType.simpleString()
        for field in source_df.schema.fields
    }

    if target_schema == source_schema:

        print("PASS: Both DataFrames are compatible for MERGE")

        return True

    else:

        print("FAIL: DataFrames are not compatible")

        print("\nTarget Schema:")
        print(target_schema)

        print("\nSource Schema:")
        print(source_schema)

        return False
```
